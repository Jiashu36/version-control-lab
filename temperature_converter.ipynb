# Temperature Converter

This program asks for your name and a temperature in Celsius, then converts the temperature to Fahrenheit and displays a personalized message.

Run the code cell below and answer both prompts. Enter only a number for the temperature.

Formula: Fahrenheit = Celsius × 9 / 5 + 32.

```python
# Get the user's name and a temperature in Celsius.
name = input("What is your name? ")

try:
    celsius = float(input("Enter a temperature in Celsius: "))
    fahrenheit = celsius * 9 / 5 + 32
    print(f"Hi, {name}! {celsius:.1f} degrees Celsius is {fahrenheit:.1f} degrees Fahrenheit.")
except ValueError:
    print("Please enter a number for the temperature, such as 25 or -10.")
```
